```python
import pandas as pd

# List of company stock files to load
files = ['aapl.csv', 'amzn.csv', 'nflx.csv', 'fb.csv', 'goog.csv']

# Dictionary to store dataframes, with company names as keys
stock_dfs = {}

# Loop through each file, read it into a dataframe, and store it
for file in files:
    company_name = file.split('.')[0]  # Extract company name from filename
    file_path = f'/content/{file}'
    try:
        df = pd.read_csv(file_path)
        stock_dfs[company_name] = df
        print(f"Successfully loaded {file} into stock_dfs['{company_name}']")
        print(f"First 5 rows of {company_name}:")
        display(df.head())
        print("\n" + "-"*50 + "\n") # Separator for readability
    except FileNotFoundError:
        print(f"Error: File not found at {file_path}")
    except Exception as e:
        print(f"Error loading {file}: {e}")
```

```python
for company_name, df in stock_dfs.items():
    df['ticker'] = company_name.upper()
    print(f"Added 'ticker' column with value '{company_name.upper()}' to {company_name} DataFrame.")
    display(df.head())
    print("\n" + "-"*50 + "\n")
```

```python
# Concatenate all dataframes into a single dataframe
all_stocks_df = pd.concat(stock_dfs.values(), ignore_index=True)

print("Successfully concatenated all stock data into 'all_stocks_df'.")
print("First 5 rows of the combined DataFrame:")
display(all_stocks_df.head())
```

```python
# Save the combined DataFrame to a CSV file
output_filename = 'faanf.csv'
all_stocks_df.to_csv(output_filename, index=False)

print(f"Combined stock data saved to '{output_filename}' successfully.")
```

```python
# Convert 'date' column to datetime
all_stocks_df['date'] = pd.to_datetime(all_stocks_df['date'])

# Convert 'volume' column to integer
all_stocks_df['volume'] = all_stocks_df['volume'].astype(int)

# Sort the DataFrame by 'date' and 'ticker'
all_stocks_df = all_stocks_df.sort_values(by=['date', 'ticker']).reset_index(drop=True)

print("Data types after conversion:")
display(all_stocks_df.info())

print("\nFirst 5 rows of the sorted DataFrame:")
display(all_stocks_df.head())
```

```python
# Sort the DataFrame by 'volume' in descending order and select the top 7 rows
highest_volume_rows = all_stocks_df.sort_values(by='volume', ascending=False).head(7)

print("Seven rows with the highest volume:")
display(highest_volume_rows)
```

```python
# Use the melt() method to reshape the DataFrame
# 'date' and 'ticker' are kept as identifier variables (id_vars)
# The remaining columns ('open', 'high', 'low', 'close', 'volume') are melted
melted_df = pd.melt(all_stocks_df, id_vars=['date', 'ticker'],
                    var_name='metric', value_name='value')

print("First 5 rows of the melted DataFrame:")
display(melted_df.head())

print("\nLast 5 rows of the melted DataFrame:")
display(melted_df.tail())
```
