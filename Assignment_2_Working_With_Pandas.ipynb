# Task
Analyze the "InternetSales.csv" dataset to summarize internet sales by year and product category, generate summary statistics for Yearly Income and Gender by Country Regions, determine the average price of the top-selling product category, and analyze the impact of home ownership on sales. ## Load and preprocess the data

### Subtask:
Load the `InternetSales.csv` file into a pandas DataFrame and handle any initial data issues like the mixed types warning.


**Reasoning**:
Load the data into a pandas DataFrame, addressing the mixed type warning by specifying `low_memory=False`, and display the head to verify.



```python
df_sales = pd.read_csv('/content/sample_data/InternetSales.csv', encoding='latin1', sep=',', low_memory=False)
display(df_sales.head())
```

## Analyze sales by year and region

### Subtask:
Calculate and summarize the internet sales by year for each country region.


**Reasoning**:
Convert 'OrderDate' to datetime, extract the year, group by year and sales territory, sum sales amount, rename the column, and display the result.



```python
df_sales['OrderDate'] = pd.to_datetime(df_sales['OrderDate'])
df_sales['OrderYear'] = df_sales['OrderDate'].dt.year
sales_by_year_region = df_sales.groupby(['OrderYear', 'SalesTerritoryKey'])['SalesAmount'].sum().reset_index()
sales_by_year_region = sales_by_year_region.rename(columns={'SalesAmount': 'Yearly Income'})
display(sales_by_year_region)
```

## Generate summary statistics

### Subtask:
Compute summary statistics (min, max, std) for Yearly Income, and analyze sales by Gender and Country Regions.


**Reasoning**:
Compute the summary statistics for 'Yearly Income' and then group by Gender and EnglishCountryRegionName to calculate the sum of 'SalesAmount'.



```python
yearly_income_summary = sales_by_year_region.groupby('SalesTerritoryKey')['Yearly Income'].agg(['min', 'max', 'std'])
sales_by_gender_region = df_sales.groupby(['Gender', 'EnglishCountryRegionName'])['SalesAmount'].sum().reset_index()
display(yearly_income_summary)
display(sales_by_gender_region)
```

## Analyze product category sales

### Subtask:
Determine which product categories generate the most sales and calculate the average price for the top-performing category.


**Reasoning**:
Group the data by product key, calculate the sum of sales amount, sort in descending order, identify the top product key, filter the dataframe for the top product, and calculate the average sales amount for the top product. Finally, display the results.



```python
product_sales = df_sales.groupby('ProductKey')['SalesAmount'].sum()
product_sales = product_sales.sort_values(ascending=False)
top_product_key = product_sales.index[0]
top_product_df = df_sales[df_sales['ProductKey'] == top_product_key]
average_price_top_product = top_product_df['SalesAmount'].mean()

display(product_sales)
display(top_product_df)
display(average_price_top_product)
```

## Analyze the impact of home ownership on sales

### Subtask:
Investigate if there's a difference in sales based on home ownership. This might require joining with another dataset if home ownership information is not in `InternetSales.csv`.


**Reasoning**:
Examine the columns in the df_sales DataFrame to determine if there is a column indicating home ownership status.



```python
display(df_sales.columns)
```

**Reasoning**:
Based on the displayed column names, there is a 'HouseOwnerFlag' column which indicates home ownership. I will group the DataFrame by this column and calculate the sum of 'SalesAmount' for each group.



```python
sales_by_home_ownership = df_sales.groupby('HouseOwnerFlag')['SalesAmount'].sum().reset_index()
display(sales_by_home_ownership)
```
